```python
import numpy as np
```

```python
X = np.array([[0,0], [0,2], [0,4], [3,0],[3,2], [3,4]])
mu1 = np.array([0, 4])
mu2 = np.array([3, 0])
d1 = X - mu1
d2 = X - mu2
s1 = np.sum(d1 ** 2, axis=1)
s2 = np.sum(d2 ** 2, axis=1)
print(s1, s2)
```

```python
labels = (s2 < s1).astype(int)
print(labels)
```

```python
X1 = X[labels==0]
X2 = X[labels==1]
mu1, mu2 = np.mean(X1, axis=0), np.mean(X2, axis=0)
print(mu1, mu2)
```

```python
d1, d2 = X - mu1, X- mu2
s1, s2 = np.sum(d1 ** 2, axis=1), np.sum(d2 ** 2, axis=1)
print(s1, s2)
```

```python
labels = (s2 < s1).astype(int)
print(labels)
```

```python
print(np.minimum(s1, s2).sum())
```
