```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def four_blobs():
    rng = np.random.default_rng(7)
    centers = np.array([[0, 0], [10, 0], [0, 10], [10, 10]], float)
    return np.vstack([c + rng.normal(0, 1.0, (40, 2)) for c in centers])
```

```python
X = four_blobs()

fig, ax = plt.subplots(1, 1, figsize=(8, 8))
ax.scatter(X[:, 0], X[:, 1], s=20, alpha=0.7, label="points")
ax.set_aspect(
    "equal"
)  # both axes are real data, so equal units keep the stretch honest
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.set_title("Four blobs data")
ax.legend()
plt.show()
```

```python
def kpp(X, K, rng):
    m = len(X)
    idx = [rng.integers(m)]
    D2 = ((X - X[idx[0]]) ** 2).sum(1)
    for _ in range(K - 1):
        j = rng.choice(m, p=D2 / D2.sum())
        idx.append(j)
        D2 = np.minimum(D2, ((X - X[j]) ** 2).sum(1))
    return X[idx].copy()
```

```python
# 3-d Broadcasting part -- To be understood
def lloyd(X, mu):
    mu = mu.copy()
    c_old = None
    for _ in range(500):
        c = ((X[:, None] - mu[None]) ** 2).sum(-1).argmin(1)
        if c_old is not None and (c == c_old).all():
            break
        c_old = c
        for k in range(len(mu)):
            if (c == k).any():
                mu[k] = X[c == k].mean(0)
    return mu, c, ((X - mu[c]) ** 2).sum(-1).mean()
```

```python
def best_J(X, K, runs=10):
    return min(lloyd(X, kpp(X, K, np.random.default_rng(s)))[2] for s in range(runs))
```

```python
print([round(best_J(X, K), 2) for K in range(1, 11)])
print("K=m J", round(best_J(X[:20], 20, 1), 4))
```
