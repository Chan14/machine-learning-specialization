```python
def kruskal(n, edges):
    """n points labeled 0..n-1; edges = [(u, v, weight), ...]
    Returns the MST as a list of edges."""
    parent = list(range(n))  # each point starts as its own component
    rank = [0] * n

    def find(x):
        # follow parents to the root; compress path along the way
        while parent[x] != x:
            parent[x] = parent[parent[x]]
            x = parent[x]
        return x

    def union(a, b):
        ra, rb = find(a), find(b)
        if ra == rb:
            return False  # same component -> edge would make a loop
        if rank[ra] < rank[rb]:
            ra, rb = rb, ra
        parent[rb] = ra  # attach shorter tree under taller
        if rank[ra] == rank[rb]:
            rank[ra] += 1
        return True

    mst = []
    # Invariant: mst connects points without loops, using the
    # shortest edges seen so far that joined different components.
    for u, v, w in sorted(edges, key=lambda e: e[2]):
        if union(u, v):
            mst.append((u, v, w))
            if len(mst) == n - 1:  # everything connected -> done
                break
    return mst, parent, rank


# Your example: 4 small blobs (pairs) -> 2 blobs -> 1
edges = [
    (0, 1, 0.5),
    (2, 3, 0.5),
    (4, 5, 0.5),
    (6, 7, 0.5),  # inside small blobs
    (1, 2, 2),
    (5, 6, 2),  # small blobs merge
    (3, 4, 10),  # the bridge
    (0, 3, 2.5),
    (0, 7, 12),  # extra edges -> skipped
]
mst, p, r = kruskal(8, edges)
print(f"parents = {p}")
print(f"rank = {r}")
for _ in mst:
    print(_)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn.datasets as data
%matplotlib inline
sns.set_context('poster')
sns.set_style('white')
sns.set_color_codes()
plot_kwds = {'alpha' : 0.5, 's' : 80, 'linewidths':0}
```

```python
moons, _ = data.make_moons(n_samples=50, noise=0.05)
blobs, _ = data.make_blobs(n_samples=50, centers=[(-0.75,2.25), (1.0, 2.0)], cluster_std=0.25)
test_data = np.vstack([moons, blobs])
plt.scatter(test_data.T[0], test_data.T[1], color='b', **plot_kwds)
```

```python
import hdbscan
```

```python
clusterer = hdbscan.HDBSCAN(min_cluster_size=5, gen_min_span_tree=True)
clusterer.fit(test_data)
```

```python
clusterer.minimum_spanning_tree_.plot(edge_cmap='viridis',
                                      edge_alpha=0.6,
                                      node_size=80,
                                      edge_linewidth=2)
```
