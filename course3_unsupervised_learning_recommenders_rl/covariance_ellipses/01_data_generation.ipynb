```python
import numpy as np
import matplotlib.pyplot as plt
%load_ext autoreload
%autoreload 2
```

```python
rng = np.random.default_rng(42)
a = rng.normal(size=3)
b = rng.normal(size=3)
print(a, b)

rng2 = np.random.default_rng(42)
c = rng2.normal(size=3)
print(c)
```

```python
# Shared generator: data and init come from one stream
for n in (100, 101):
    rng = np.random.default_rng(42)
    X = rng.normal(size=(n, 2))  # data uses up 2n numbers
    init = rng.integers(0, n, size=3)  # init draws whatever comes next
    print(init)
```

```python
# Separate generators: one stream per job
for n in (100, 101):
    rng_data = np.random.default_rng(0)
    rng_init = np.random.default_rng(1)
    X = rng_data.normal(size=(n, 2))
    init = rng_init.integers(0, 100, size=3)  # identical call both times
    print(init)
```

```python
root = np.random.default_rng(42)
rng_data, rng_init = root.spawn(2)
X = rng_data.normal(size=(4, 2))
init = rng_init.integers(0, 10, size=3)
print(X, init)
rng_extra = root.spawn(1)[0]  # third child. spawning doesn't consume root's draws.
```

```python
rng_data, rng_init = np.random.default_rng(np.random.SeedSequence(42)).spawn(2)
rng_extra = np.random.default_rng(
    np.random.SeedSequence(
        42,
        spawn_key=(1,),
    )
)
X = rng_data.normal(size=(4, 2))
init = rng_init.integers(0, 10, size=3)
init_extra = rng_extra.integers(0, 10, size=3)
print(X, init, init_extra)
```

```python
root = np.random.default_rng(0)
rng_data, rng_jitter = root.spawn(2)
n = 1000

A = rng_data.normal(loc=0.0, scale=1.0, size=n)
jitter = rng_jitter.uniform(-0.5, 0.5, size=n)

fig, (ax_hist, ax_strip) = plt.subplots(
    2, 1, figsize=(8, 4), sharex=True, height_ratios=[3, 1]
)

ax_hist.hist(A, bins=40, density=True, alpha=0.7, edgecolor="white")
ax_hist.set_ylabel("density")
ax_hist.set_title("1D Gaussian: histogram + jittered strip")

ax_strip.scatter(A, jitter, s=6, alpha=0.3)
ax_strip.set_yticks([])
ax_strip.set_xlabel("value")

plt.tight_layout()
plt.show()
```

```python
rng = np.random.default_rng(0)
n = 1000

B = rng.normal(
    loc=[0.0, 5.0], scale=[1.0, 3.0], size=(n, 2)
)  # shape (n, 2): rows = points, cols = (x, y)
mu = B.mean(axis=0)
label = f"Mean ({mu[0]:.2f}, {mu[1]:.2f})"
fig, ax = plt.subplots(figsize=(8, 8))
ax.scatter(B[:, 0], B[:, 1], s=8, alpha=0.3)
ax.scatter(mu[0], mu[1], c="red", marker="x", s=40, label=label)
ax.set_aspect(
    "equal"
)  # both axes are real data, so equal units keep the stretch honest
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.set_title("B: x ~ N(0, 1²), y ~ N(5, 3²)")
ax.grid(alpha=0.4)
ax.legend()
plt.show()
```

```python
rng = np.random.default_rng(0)
n = 1000

C = rng.uniform(low=-2.0, high=2.0, size=(n, 2))
mu = C.mean(axis=0)
mu_label = f"Sample Mean ({mu[0]:.2f}, {mu[1]:.2f})"
print(mu)

fig, ax = plt.subplots(figsize=(8, 8))
ax.scatter(C[:, 0], C[:, 1], s=8, alpha=0.7)
ax.scatter(mu[0], mu[1], c="red", marker="x", s=40, label=mu_label)
ax.set_xlim(-2.5, 2.5)
ax.set_ylim(-2.5, 2.5)
ax.set_aspect(
    "equal"
)  # both axes are real data, so equal units keep the stretch honest
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.set_title("placeholder")
ax.grid(alpha=0.4)
ax.legend()
plt.show()
```

```python
rng = np.random.default_rng(0)
n = 1000
low, high = -2.0, 2.0

C = rng.uniform(
    low=low, high=high, size=(n, 2)
)  # shape (n, 2): rows = points, cols = (x, y)

data = C  # single source for every panel below
mu = data.mean(axis=0)
sigma = data.std(axis=0)
sigma_theory = (high - low) / np.sqrt(12)
print(f"mean = {mu.round(3)}, std = {sigma.round(3)}, theory std = {sigma_theory:.3f}")

fig = plt.figure(figsize=(8, 8))
gs = fig.add_gridspec(
    2, 2, width_ratios=[4, 1], height_ratios=[1, 4], hspace=0.05, wspace=0.05
)
ax = fig.add_subplot(gs[1, 0])
ax_top = fig.add_subplot(gs[0, 0], sharex=ax)
ax_side = fig.add_subplot(gs[1, 1], sharey=ax)

# main scatter
ax.scatter(data[:, 0], data[:, 1], s=8, alpha=0.5)
ax.scatter(
    *mu, c="red", marker="x", s=120, label=f"sample mean ({mu[0]:.2f}, {mu[1]:.2f})"
)
# ax.add_patch(plt.Rectangle((low, low), high - low, high - low,
# fill=False, ls="--", lw=1.5, ec="black", label="support [-2, 2]²"))
pad = 0.5
ax.set_xlim(low - pad, high + pad)
ax.set_ylim(low - pad, high + pad)
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.grid(alpha=0.3)
ax.legend(loc="upper right", fontsize=9)

# marginals: should be flat at height 1 / (high - low) = 0.25
flat = 1 / (high - low)
ax_top.hist(data[:, 0], bins=40, density=True, alpha=0.7)
ax_top.hlines(flat, low, high, color="red", ls="--", lw=1)  # only from -2 to 2
ax_side.hist(data[:, 1], bins=40, density=True, alpha=0.7, orientation="horizontal")
ax_side.vlines(flat, low, high, color="red", ls="--", lw=1)
ax_top.tick_params(labelbottom=False)
ax_side.tick_params(labelleft=False)

fig.suptitle("C: x, y ~ Uniform(-2, 2), independent", y=0.95)
plt.show()
```
