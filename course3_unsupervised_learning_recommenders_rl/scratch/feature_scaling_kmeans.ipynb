```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

rng = np.random.default_rng(0)
n = 300

# Case A (scaling helps): same income spread, children ~1 vs ~2
A0 = np.c_[rng.normal(60_000, 20_000, n), rng.normal(1, 0.15, n)]
A1 = np.c_[rng.normal(60_000, 20_000, n), rng.normal(2, 0.15, n)]

# Case B (scaling hurts): income 30k vs 90k, irrelevant 0/1 "has pet"
B0 = np.c_[rng.normal(30_000, 10_000, n), rng.integers(0, 2, n)]
B1 = np.c_[rng.normal(90_000, 10_000, n), rng.integers(0, 2, n)]

truth = np.r_[np.zeros(n), np.ones(n)]


def accuracy(labels):
    # k-means label numbers are arbitrary, so take the better of the two matchings
    return max((labels == truth).mean(), (labels != truth).mean())


fig, axs = plt.subplots(2, 2, figsize=(13, 10))
cases = [
    ("A: income vs children", A0, A1, "children"),
    ("B: income vs has_pet", B0, B1, "has_pet"),
]

for row, (name, G0, G1, ylab) in enumerate(cases):
    X = np.vstack([G0, G1])
    Z = StandardScaler().fit_transform(X)  # per-column (x - mean) / std
    for col, (D, tag) in enumerate([(X, "raw"), (Z, "scaled")]):
        labels = KMeans(2, n_init=10, random_state=0).fit_predict(D)
        ax = axs[row, col]
        ax.scatter(D[:, 0], D[:, 1], c=labels, cmap="coolwarm", s=8, alpha=0.5)
        ax.set_title(f"{name} — {tag}\nk-means accuracy {accuracy(labels):.0%}")
        ax.set_xlabel("income" + (" (z)" if tag == "scaled" else ""))
        ax.set_ylabel(ylab + (" (z)" if tag == "scaled" else ""))
        ax.grid(alpha=0.4)
        if tag == "scaled":
            ax.set_aspect(
                "equal"
            )  # equal units after scaling, so distances look honest

plt.tight_layout()
plt.show()
```

```python
from collections import defaultdict

rng_p1 = np.random.default_rng(47)
n = 10000
rolls = np.array([1, 2, 3, 4, 5, 6])
probs1 = np.array([1, 1, 1, 1, 1, 1]) / 6
probs2 = np.array([1, 3, 5, 5, 3, 1]) / 18
picks1, picks2 = defaultdict(int), defaultdict(int)
for _ in range(n):
    pick1, pick2 = rng_p1.choice(rolls, p=probs1), rng_p1.choice(rolls, p=probs2)
    picks1[pick1] += 1
    picks2[pick2] += 1
print("Case 1 - Equal probs")
total = 0
for k, v in sorted(picks1.items()):
    print(f"{k}: {v/n:.2f}")
    total += k * v
print(f"Weighted average: {total/n}")

total = 0
print("Case 2 - Different probs")
for k, v in sorted(picks2.items()):
    print(f"{k}: {v/n:.2f}")
    total += k * v
print(f"Weighted average: {total/n}")
```

```python
rng = np.random.default_rng(47)
faces = np.arange(1, 7)
probs = np.array([1, 3, 5, 5, 3, 1]) / 18
# probs = np.array([1, 1, 1, 1, 1, 1]) / 6

draws = rng.choice(faces, size=10_000, p=probs)  # all rolls at once
freq = np.bincount(draws, minlength=7)[1:] / draws.size  # index 0 unused

print(np.round(freq, 4))  # compare with probs
print(draws.mean(), faces @ probs)  # simulated vs expected mean
```
